```python
#Define the custom ranking for each ordinal variable
airline_categories = ["Jet Airways", "IndiGo", "Air India","Multiple carriers","SpiceJet","Vistara","Air Asia","GoAir","Multiple carriers Premium economy","Jet Airways Business","Vistara Premium economy","Trujet"]
source_categories = ["Delhi","Kolkata","Banglore","Mumbai","Chennai"]
destination_categories = ["Cochin","Banglore","Delhi","New Delhi","Hyderabad","Kolkata"]
```

```python
#Date_of_journey --> Day, Month
X["Date_of_Journey"] = pd.to_datetime(
    X["Date_of_Journey"],
    format = "%d/%m/%Y"
)
X["Journey_Day"] = X["Date_of_Journey"].dt.day
X["Journey_Month"] = X["Date_of_Journey"].dt.month

#Dep_time --> Hour, Minute
# Dep_Time → Hour, Minute
X["Dep_Time"] = pd.to_datetime(
    X["Dep_Time"],
    format="%H:%M"
)

X["Dep_Hour"] = X["Dep_Time"].dt.hour
X["Dep_Minute"] = X["Dep_Time"].dt.minute
```

```python
X.head()
```

```python
X.info()
```

```python
X = X.drop(["Date_of_Journey","Dep_Time"],axis=1)
```

```python
X.head()
```

```python
categorical_cols = X.select_dtypes(include="object").columns
```

```python
numerical_cols = X.select_dtypes(exclude="object").columns
```

```python
from sklearn.impute import SimpleImputer ## HAndling Missing Values
from sklearn.preprocessing import StandardScaler # HAndling Feature Scaling
from sklearn.preprocessing import OrdinalEncoder # Ordinal Encoding
## pipelines
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
categorical_cols
```

```python
numerical_cols
```

```python
number_pipeline = Pipeline(
    steps = [
        ('imputer',SimpleImputer()),
        ('scaler',StandardScaler())
    ]
)
```

```python
cat_pipeline = Pipeline(
    steps =[
        ('imputer',SimpleImputer(strategy='most_frequent')),
        ('ordinalencoder',OrdinalEncoder(categories=[airline_categories,source_categories,destination_categories]))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    [
        ('num_pipeline',number_pipeline, numerical_cols),
        ('cat_pipeline',cat_pipeline,categorical_cols)
    ]
)
```

```python
## Train test split

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
preprocessor.fit_transform(X_train)
```
