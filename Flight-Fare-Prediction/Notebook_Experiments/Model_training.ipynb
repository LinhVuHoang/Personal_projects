```python
#Date_of_journey --> Day, Month
X["Date_of_Journey"] = pd.to_datetime(
    X["Date_of_Journey"],
    format = "%d/%m/%Y"
)
X["Journey_Day"] = X["Date_of_Journey"].dt.day
X["Journey_Month"] = X["Date_of_Journey"].dt.month

#Dep_time --> Hour, Minute
# Dep_Time → Hour, Minute
X["Dep_Time"] = pd.to_datetime(
    X["Dep_Time"],
    format="%H:%M"
)

X["Dep_Hour"] = X["Dep_Time"].dt.hour
X["Dep_Minute"] = X["Dep_Time"].dt.minute
```

```python
X = X.drop(["Date_of_Journey","Dep_Time"],axis=1)
```

```python
categorical_cols = X.select_dtypes(include="object").columns
```

```python
numerical_cols = X.select_dtypes(exclude="object").columns
```

```python
from sklearn.impute import SimpleImputer ## HAndling Missing Values
from sklearn.preprocessing import StandardScaler # HAndling Feature Scaling
from sklearn.preprocessing import OrdinalEncoder # Ordinal Encoding
from sklearn.preprocessing import OneHotEncoder #Onehot Encoding
## pipelines
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
categorical_cols
```

```python
numerical_cols
```

```python
number_pipeline = Pipeline(
    steps = [
        ('imputer',SimpleImputer(strategy='median')),
        ('scaler',StandardScaler())
    ]
)
```

```python
cat_pipeline = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehotencoder', OneHotEncoder(
            handle_unknown='ignore',
            sparse_output=False
        ))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    [
        ('num_pipeline',number_pipeline, numerical_cols),
        ('cat_pipeline',cat_pipeline,categorical_cols)
    ]
)
```

```python
## Train test split

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
preprocessor.fit_transform(X_train)[0]
```

```python
preprocessor.transform(X_test)
```

```python
x_train = preprocessor.fit_transform(X_train)
x_test = preprocessor.transform(X_test)
x_train = pd.DataFrame(x_train)
x_test = pd.DataFrame(x_test)
```

```python
#Build the regression / Regressor models

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn import linear_model
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
import xgboost as xgb
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.ensemble import BaggingRegressor
from sklearn.ensemble import GradientBoostingRegressor

#Create objects of Regression/ Regressor models with default hyper-parameter
modelmlg = LinearRegression()
modeldcr = DecisionTreeRegressor()
modelbag = BaggingRegressor()
modelrfr = RandomForestRegressor()
modelSVR = SVR()
modelXGR = xgb.XGBRegressor()
modelKNN = KNeighborsRegressor(n_neighbors=5)
modelETR = ExtraTreesRegressor()
modelRE=Ridge()
modelLO=linear_model.Lasso(alpha=0.1)

modelGBR = GradientBoostingRegressor(loss='squared_error', learning_rate=0.1, n_estimators=100, subsample=1.0,
                                     criterion='friedman_mse', min_samples_split=2, min_samples_leaf=1,
                                     min_weight_fraction_leaf=0.0, max_depth=3, min_impurity_decrease=0.0,
                                     init=None, random_state=None, max_features=None,
                                     alpha=0.9, verbose=0, max_leaf_nodes=None, warm_start=False,
                                     validation_fraction=0.1, n_iter_no_change=None, tol=0.0001, ccp_alpha=0.0)

```

```python
# Evalution matrix for all the algorithms
MM = [modelmlg, modeldcr, modelrfr, modelKNN, modelETR, modelGBR, modelXGR, modelbag,modelRE,modelLO]
```

```python
results =[]
for models in MM:
    
    #Fit the model with train data
    models.fit(x_train, y_train)
    
    #Predict the model with test data
    
    y_pred = models.predict(x_test)
    
    #Print the model name
    
    print("Model name: ",models)
    
    #Evaluation metrics for Regression analysis
    
    from sklearn import metrics #Error (càng thấp càng tốt), score (Càng cao càng tốt)
    print('Mean Absolute Error (MAE):', round(metrics.mean_absolute_error(y_test, y_pred),3))  
    print('Mean Squared Error (MSE):', round(metrics.mean_squared_error(y_test, y_pred),3))  
    print('Root Mean Squared Error (RMSE):', round(np.sqrt(metrics.mean_squared_error(y_test, y_pred)),3))
    print('R2_score:', round(metrics.r2_score(y_test, y_pred),6))
    print('Root Mean Squared Log Error (RMSLE):', round(np.log(np.sqrt(metrics.mean_squared_error(y_test, y_pred))),3))
    # Define the function to calculate the MAPE - Mean Absolute Percentage Error

    def MAPE (y_test, y_pred):
        y_test, y_pred = np.array(y_test), np.array(y_pred)
        return np.mean(np.abs((y_test - y_pred) / y_test)) * 100
    
    # Evaluation of MAPE 

    result = MAPE(y_test, y_pred)
    print('Mean Absolute Percentage Error (MAPE):', round(result, 2), '%')
     # Evaluation of MAPE 

    result = MAPE(y_test, y_pred)
    print('Mean Absolute Percentage Error (MAPE):', round(result, 2), '%')
    
    # Calculate Adjusted R squared values 

    r_squared = round(metrics.r2_score(y_test, y_pred),6)
    adjusted_r_squared = round(1 - (1-r_squared)*(len(y)-1)/(len(y)-X.shape[1]-1),6)
    print('Adj R Square: ', adjusted_r_squared)
    
    #Save result
    
    results.append({
        "Model": models,
        'MAE': round(metrics.mean_absolute_error(y_test, y_pred),3),
        'MSE': round(metrics.mean_squared_error(y_test, y_pred),3),
        'RMSE': round(np.sqrt(metrics.mean_squared_error(y_test, y_pred)),3),
        'R2_score': round(metrics.r2_score(y_test, y_pred),6),
        'RMSLE': round(np.log(np.sqrt(metrics.mean_squared_error(y_test, y_pred))),3)
    })
    print('------------------------------------------------------------------------------------------------------------')
```
